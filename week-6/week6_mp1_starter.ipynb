## Section 4 — Visualization

Create at least one visualization that supports one of your analysis findings. Your chart should:

- Have a title that states the finding, not just the data (e.g., "Satisfaction scores drop sharply after age 40" not "Satisfaction by age")
- Have labeled axes
- Use a chart type appropriate for your data (bar for categorical comparison, scatter for relationships, line for trends over time)

Below the chart, explain in a markdown cell: why you chose this chart type, and what you want the reader to take away from it.

```python
# Your visualization
# 1) Histograms: sugars_100g, fiber_100g, energy-kcal_100g
fig = px.histogram(
    df,
    x="sugars_100g",
    nbins=30,
    title="Distribution of Sugar (g/100g)",
    labels={"sugars_100g": "Sugar (g/100g)", "count": "Number of cereals"},
)
fig.show()

fig = px.histogram(
    df,
    x="fiber_100g",
    nbins=30,
    title="Distribution of Fiber (g/100g)",
    labels={"fiber_100g": "Fiber (g/100g)", "count": "Number of cereals"},
)
fig.show()

fig = px.histogram(
    df,
    x="energy-kcal_100g",
    nbins=30,
    title="Distribution of Energy (kcal/100g)",
    labels={"energy-kcal_100g": "Energy (kcal/100g)", "count": "Number of cereals"},
)
fig.show()
```

```python
# 2) Box plots by nutrition grade
grade_order = ["a", "b", "c", "d", "e", "unknown"]

fig = px.box(
    df,
    x="nutrition_grade_fr",
    y="sugars_100g",
    category_orders={"nutrition_grade_fr": grade_order},
    title="Sugar by Nutrition Grade",
    labels={"nutrition_grade_fr": "Nutrition Grade", "sugars_100g": "Sugar (g/100g)"},
    points="outliers",
)
fig.show()

fig = px.box(
    df,
    x="nutrition_grade_fr",
    y="fiber_100g",
    category_orders={"nutrition_grade_fr": grade_order},
    title="Fiber by Nutrition Grade",
    labels={"nutrition_grade_fr": "Nutrition Grade", "fiber_100g": "Fiber (g/100g)"},
    points="outliers",
)
fig.show()

fig = px.box(
    df,
    x="nutrition_grade_fr",
    y="energy-kcal_100g",
    category_orders={"nutrition_grade_fr": grade_order},
    title="Energy by Nutrition Grade",
    labels={"nutrition_grade_fr": "Nutrition Grade", "energy-kcal_100g": "Energy (kcal/100g)"},
    points="outliers",
)
fig.show()
```

```python
# 3) Scatter plot (sugar vs fiber, colored by nutrition grade)
fig = px.scatter(
    df,
    x="sugars_100g",
    y="fiber_100g",
    color="nutrition_grade_fr",
    hover_data=["product_name", "brands", "countries", "energy-kcal_100g"],
    title="Sugar vs Fiber by Nutrition Grade",
    labels={"sugars_100g": "Sugar (g/100g)", "fiber_100g": "Fiber (g/100g)"},
    opacity=0.75,
)
fig.show()

# Optional second scatter: sugar vs energy
fig = px.scatter(
    df,
    x="sugars_100g",
    y="energy-kcal_100g",
    color="nutrition_grade_fr",
    hover_data=["product_name", "brands", "countries", "fiber_100g"],
    title="Sugar vs Energy by Nutrition Grade",
    labels={"sugars_100g": "Sugar (g/100g)", "energy-kcal_100g": "Energy (kcal/100g)"},
    opacity=0.75,
)
fig.show()
```

```python
# 4) Top-N brand comparison bar chart (mean values)
top_n = 10
brand_stats = (
    df.groupby("brands", dropna=False)
      .agg(
          n_products=("product_name", "count"),
          mean_sugar=("sugars_100g", "mean"),
          mean_fiber=("fiber_100g", "mean"),
          mean_energy=("energy-kcal_100g", "mean"),
      )
      .reset_index()
)

top_brands = (
    brand_stats[brand_stats["brands"].notna()]
    .sort_values("n_products", ascending=False)
    .head(top_n)
)

fig = px.bar(
    top_brands.sort_values("mean_sugar", ascending=False),
    x="brands",
    y="mean_sugar",
    color="n_products",
    title=f"Top {top_n} Brands: Average Sugar",
    labels={"brands": "Brand", "mean_sugar": "Avg Sugar (g/100g)", "n_products": "# Products"},
)
fig.update_layout(xaxis_tickangle=-35)
fig.show()

fig = px.bar(
    top_brands.sort_values("mean_energy", ascending=False),
    x="brands",
    y="mean_energy",
    color="n_products",
    title=f"Top {top_n} Brands: Average Energy",
    labels={"brands": "Brand", "mean_energy": "Avg Energy (kcal/100g)", "n_products": "# Products"},
)
fig.update_layout(xaxis_tickangle=-35)
fig.show()
```
